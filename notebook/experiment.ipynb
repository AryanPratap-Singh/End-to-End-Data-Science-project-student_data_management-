### total score

```python
df["total_score"] = df["math_score"] + df["reading_score"] + df["writing_score"]
df["average"]=df["total_score"]/3
df.head()
```

### exploring data

histogram for average

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 7))

sns.histplot(data=df, x="average", kde=True, bins=30, color="steelblue", ax=axes[0])
axes[0].set_title("Math Score Distribution")
axes[0].set_xlabel("Math score")
axes[0].set_ylabel("Number of students")

sns.histplot(data=df, x="average",kde=True, hue="gender", ax=axes[1])
axes[1].set_title("Math Score Boxplot")
axes[1].set_xlabel("Math score")

fig.suptitle("Math Score Analysis", fontsize=14)
fig.tight_layout()
```

histogram for total

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 7))

sns.histplot(data=df, x="total_score", kde=True, bins=30, color="steelblue", ax=axes[0])

sns.histplot(data=df, x="total_score",kde=True, hue="gender", ax=axes[1])

fig.suptitle("Math Score Analysis", fontsize=14)
fig.tight_layout()
```
